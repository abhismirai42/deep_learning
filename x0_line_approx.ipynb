```python
import numpy as np
import matplotlib.pyplot as plt

X = np.linspace(0, 1, 10)
W, B = 2, 1
Y = W * X + B

#lets do some learning
w, b = 0, 0 #initial guess

#learning rate
alpha = 0.1

for i in range(50):
    #forward pass
    y_pred = w * X + b

    #loss
    loss = (y_pred - Y) ** 2
    total_loss = loss.sum()
    print("total loss: ", total_loss)

    dl_dw = ((y_pred - Y) * X).sum()
    dl_db = (y_pred - Y).sum()



    w = w - alpha * dl_dw
    b = b - alpha * dl_db

    if (i%10 == 0):
        plt.plot(X, y_pred, label = "pred")
        plt.plot(X, Y, label = "data")
        plt.legend()
        plt.show()
```
